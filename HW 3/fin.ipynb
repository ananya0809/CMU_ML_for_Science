```python
TEST_X = "f24-38615-hw3/test_X.csv"
TRAIN_X = "f24-38615-hw3/train_X.csv"
TRAIN_Y = "f24-38615-hw3/train_y.csv"
```

```python
import pandas as pd
```

```python
x = pd.read_csv(TRAIN_X).drop(columns=['Unnamed: 0'])
```

```python
y = pd.read_csv(TRAIN_Y).drop(columns=['Unnamed: 0'])
```

```python
tt_x_or = pd.read_csv(TEST_X)
tt_x = tt_x_or.drop(columns=['Unnamed: 0'])
```

```python
import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LassoCV, LogisticRegression
from sklearn.metrics import accuracy_score

# Assuming X and Y are your data arrays
X_train, X_test, Y_train, Y_test = train_test_split(x, y, test_size=0.3, random_state=42)

# Step 1: Standardize the data
# scaler = StandardScaler()
# X_train_scaled = scaler.fit_transform(X_train)
# X_test_scaled = scaler.transform(X_test)
```

```python

# Step 2: Apply LASSO for feature selection
lasso = LassoCV(cv=10, random_state=42, max_iter=10000000)
lasso.fit(X_train, Y_train)
```

```python
sorte = np.argsort(np.abs(lasso.coef_))
def accuracy_check(idx) :
# Get indices of non-zero coefficients
    important_features_idx = sorte[idx:idx + 1]

    X_train_reduced = X_train.iloc[:, important_features_idx]
    X_test_reduced = X_test.iloc[:, important_features_idx]

    # Step 4: Logistic Regression on selected features
    log_reg = LogisticRegression(penalty='l2', max_iter=1000000)
    log_reg.fit(X_train_reduced, Y_train)

    # Step 5: Predict and check accuracy
    Y_pred = log_reg.predict(X_test_reduced)
    accuracy = accuracy_score(Y_test, Y_pred)

    print(f"Model accuracy: {accuracy:.4f}")
    return accuracy
```

```python
ref = {}
for i in range(0, len(sorte)):
    ref[i] = accuracy_check(i)
```

```python
hmm = sorted(ref.items(), key=lambda x:x[1])
hmm.reverse()
```

```python
from sklearn.model_selection import GridSearchCV
from sklearn.linear_model import Lasso
from sklearn.linear_model import LogisticRegression

# Step 1: Hyperparameter tuning for LASSO
lasso_params = {'alpha': [0.001, 0.01, 0.1, 1, 10, 100]}
lasso_grid = GridSearchCV(Lasso(max_iter=10000), param_grid=lasso_params, cv=5)
lasso_grid.fit(X_train, Y_train)

best_lasso = lasso_grid.best_estimator_
important_features_idx = np.argsort(np.abs(best_lasso.coef_))[-15000:]

X_train_reduced = X_train.iloc[:, important_features_idx]
X_test_reduced = X_test.iloc[:, important_features_idx]

# Step 2: Hyperparameter tuning for Logistic Regression
log_reg_params = {'C': [0.001, 0.01, 0.1, 1, 10, 100], 'penalty': ['l2']}
log_reg_grid = GridSearchCV(LogisticRegression(max_iter=10000), param_grid=log_reg_params, cv=5)
log_reg_grid.fit(X_train_reduced, Y_train)

best_log_reg = log_reg_grid.best_estimator_

# Step 3: Evaluate accuracy
Y_pred = best_log_reg.predict(X_test_reduced)
accuracy = accuracy_score(Y_test, Y_pred)
print(f"Model accuracy after tuning: {accuracy:.4f}")
```
