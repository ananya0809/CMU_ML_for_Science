```python
# import required libraries
import pandas as pd
from sklearn.decomposition import PCA
from sklearn.preprocessing import StandardScaler
```

```python
# read dataset csv
hw = pd.read_csv('hw1_dataset.csv')
hw
```

```python
# X = features, Y = target
# features in X are contributing in some manner to result in the Y value
data_X = hw.iloc[:,1:]
data_Y = hw.iloc[:, 0]
data_X
data_Y
```

```python
# the dataset has columns that contain string values, we only filter out the numerical columns
# also remove any rows that might have null or NA values
cleaned_X = data_X.select_dtypes(include=['int64', 'float64']).dropna()

# modify Y to remove rows that were removed from X
cleaned_Y = data_Y.loc[cleaned_X.index].reset_index(drop=True)
cleaned_X = cleaned_X.reset_index(drop=True)
```

```python
PCA_COMPONENTS = 50
```

```python
# standardize the features (important for PCA)
scaler = StandardScaler()
scaled_X = scaler.fit_transform(cleaned_X)

# initialize PCA with components count
pca = PCA(n_components=PCA_COMPONENTS)

# fit PCA on the scaled data and transform X
pca_X = pca.fit_transform(scaled_X)

# create a df for the PCA results
df_pca_X = pd.DataFrame(pca_X, columns=[f'PC{i+1}' for i in range(PCA_COMPONENTS)])

# concatenate the PCA results with the target column from Y
final_df = pd.concat([df_pca_X, cleaned_Y.reset_index(drop=True)], axis=1)
print(final_df)
explained_variance = pca.explained_variance_ratio_
print(f'Explained variance by each component: {explained_variance}')
```

```python
pca_components = pd.DataFrame(pca.components_, columns=cleaned_X.columns, index=[f'PC{i+1}' for i in range(pca.n_components_)])
pca_components
```

```python
for i in range(1,PCA_COMPONENTS + 1):
    top_features_pcx = pca_components.loc[f'PC{i}'].abs().sort_values(ascending=False)
    print(top_features_pcx.index[0])
```
